```python
import pandas as pd
from pathlib import Path

extracted_path = Path("../data/extracted")
for csv_file in extracted_path.glob("*.csv"):
    df = pd.read_csv(csv_file)
    print(csv_file.name)
    print(f"  shape: {df.shape}, nulls: {df.isnull().sum().sum()}")
    print(f"  dtypes: {df.dtypes.value_counts().to_dict()}")
```
